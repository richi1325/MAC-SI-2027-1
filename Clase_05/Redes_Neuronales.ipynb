```python
import tensorflow as tf
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# X: [Metros cuadrados, Habitaciones, Antigüedad en años]
X = np.array([
    [80.0, 2.0, 5.0],
    [100.0, 3.0, 10.0],
    [120.0, 3.0, 2.0],
    [50.0, 1.0, 15.0],
    [200.0, 4.0, 20.0],
    [150.0, 3.0, 5.0]
], dtype=np.float32)

# y: Precio de la casa (Variable objetivo)
y = np.array([255000, 340000, 388000, 135000, 580000, 445000], dtype=np.float32)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X)
```

```python
X_norm = scaler.transform(X)
X_norm
```

```python
# Red Neuronal de Regresión Lineal
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(3,)),
    tf.keras.layers.Dense(units = 1, activation = "linear")
])
```

```python
model.compile(
    #optimizer = "sgd", # Regla de actualización
    optimizer = tf.keras.optimizers.SGD(learning_rate = 0.5),
    loss = "mse"
)
```

```python
model.fit(X_norm, y, epochs = 500, verbose = 0)
```

```python
pesos, sesgo = model.layers[0].get_weights()
```

```python
print(
    f"Precio = {pesos[0][0]} * metros_cuadrados_norm  + "
    f"{pesos[1][0]} * num_habitaciones_norm + "
    f"{pesos[2][0]} * antiguedad_años_norm + "
    f"{sesgo[0]}"
)
```

```python
nueva_casa = np.array([[150.0, 4.0, 3.0]])
# Escalar la variable al momento de predecir.
model.predict(scaler.transform(nueva_casa))
```
